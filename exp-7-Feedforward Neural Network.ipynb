```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras import regularizers
from tensorflow.keras import layers,models
from tensorflow.keras.layers import Dense,Dropout,Flatten
from tensorflow.keras.initializers import GlorotNormal,HeNormal
```

```python
(x_train,y_train),(x_test,y_test)=tf.keras.datasets.cifar10.load_data()
```

```python
x_train,x_test=x_train/255.0,x_test/255.0
```

```python
#base model
base_model=models.Sequential()
base_model.add(layers.Flatten(input_shape=(32,32,3)))
base_model.add(layers.Dense(256,activation='relu'))
base_model.add(layers.Dense(128,activation='relu'))
base_model.add(layers.Dense(10,activation='softmax'))
```

```python
xavier_model=models.Sequential()
xavier_model.add(layers.Flatten(input_shape=(32,32,3)))
xavier_model.add(layers.Dense(256,activation='relu',kernel_initializer=GlorotNormal()))
xavier_model.add(layers.Dense(128,activation='relu',kernel_initializer=GlorotNormal()))
xavier_model.add(layers.Dense(10,activation='softmax'))
```

```python
kaiming_model=models.Sequential()
kaiming_model.add(layers.Flatten(input_shape=(32,32,3)))
kaiming_model.add(layers.Dense(256,activation='relu',kernel_initializer=HeNormal()))
kaiming_model.add(layers.Dense(128,activation='relu',kernel_initializer=HeNormal()))
kaiming_model.add(layers.Dense(10,activation='softmax'))
```

```python
model_dropout=models.Sequential()
model_dropout.add(layers.Flatten(input_shape=(32,32,3)))
model_dropout.add(layers.Dense(256,activation='relu'))
model_dropout.add(Dropout(0.5))
model_dropout.add(layers.Dense(128,activation='relu'))
model_dropout.add(Dropout(0.5))
model_dropout.add(layers.Dense(10,activation='softmax'))
```

```python
model_l2=models.Sequential()
model_l2.add(layers.Flatten(input_shape=(32,32,3)))
model_l2.add(layers.Dense(256,activation='relu',kernel_regularizer=regularizers.L2(0.001)))
model_l2.add(layers.Dense(128,activation='relu',kernel_regularizer=regularizers.L2(0.001)))
model_l2.add(layers.Dense(10,activation='softmax'))
```

```python
model_l1=models.Sequential()
model_l1.add(layers.Flatten(input_shape=(32,32,3)))
model_l1.add(layers.Dense(256,activation='relu',kernel_regularizer=regularizers.L1(0.001)))
model_l1.add(layers.Dense(128,activation='relu',kernel_regularizer=regularizers.L1(0.001)))
model_l1.add(layers.Dense(10,activation='softmax'))
```

```python
base_model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
xavier_model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
kaiming_model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model_dropout.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model_l2.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model_l1.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
base_history=base_model.fit(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```

```python
xavier_history=xavier_model.fit(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```

```python
kaiming_history=kaiming_model.fit(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```

```python
dropout_history=model_dropout.fit(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```

```python
l2_history=model_l2.fit(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```

```python
l1_history=model_l1.fit(x_train,y_train,epochs=5,validation_data=(x_test,y_test))
```

```python
plt.figure(figsize=(12,4))
plt.plot(base_history.history['accuracy'],label='base training accuracy')
plt.plot(base_history.history['val_accuracy'],label='base validation accuracy')
plt.xlabel('epoch')
plt.ylabel('accuracy')
plt.legend()
plt.title('base model training and validation accuracy')
plt.show()
```

```python
plt.figure(figsize=(12,4))
plt.plot(xavier_history.history['accuracy'],label='xavier training accuracy')
plt.plot(xavier_history.history['val_accuracy'],label='xavier validation accuracy')
plt.xlabel('epoch')
plt.ylabel('accuracy')
plt.legend()
plt.title('xavier model training and validation accuracy')
plt.show()
```

```python
plt.figure(figsize=(12,4))
plt.plot(kaiming_history.history['accuracy'],label='kaiming training accuracy')
plt.plot(kaiming_history.history['val_accuracy'],label='kaiming validation accuracy')
plt.xlabel('epoch')
plt.ylabel('accuracy')
plt.legend()
plt.title('kaiming model training and validation accuracy')
plt.show()
```

```python
plt.figure(figsize=(12,4))
plt.plot(dropout_history.history['accuracy'],label='dropout training accuracy')
plt.plot(dropout_history.history['val_accuracy'],label='dropout validation accuracy')
plt.xlabel('epoch')
plt.ylabel('accuracy')
plt.legend()
plt.title('dropout model training and validation accuracy')
plt.show()
```

```python
plt.figure(figsize=(12,4))
plt.plot(l2_history.history['accuracy'],label='l2 training accuracy')
plt.plot(l2_history.history['val_accuracy'],label='l2 validation accuracy')
plt.xlabel('epoch')
plt.ylabel('accuracy')
plt.legend()
plt.title('l2 model training and validation accuracy')
plt.show()
```

```python
plt.figure(figsize=(12,4))
plt.plot(l1_history.history['accuracy'],label='l1 training accuracy')
plt.plot(l1_history.history['val_accuracy'],label='l1 validation accuracy')
plt.xlabel('epoch')
plt.ylabel('accuracy')
plt.legend()
plt.title('l1 model training and validation accuracy')
plt.show()
```
